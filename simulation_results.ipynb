```python
import matplotlib.pyplot as plt
import pandas as pd
import pickle
from pathlib import Path
```

```python
def print_result(result_dict, n_parcels):
    for i in range(len(result_dict['N'])):
        print(f"With {n_parcels} parcels, for N={result_dict['N'][i]} estimated mean sensitivty: {round(result_dict['mean_sensitivity'][i], 2)}, with mean specificity of {round(result_dict['mean_specificity'][i], 2)}.")
```

```python
pkl_path = Path("/home/neuromod/cwas_simulation")
```

```python
with open(pkl_path / 'sim_result_20parcels.pkl', 'rb') as f:
    result_20parcels = pickle.load(f)
    
with open(pkl_path / 'sim_result_36parcels.pkl', 'rb') as f:
    result_36parcels = pickle.load(f)
    
with open(pkl_path / 'sim_result_64parcels.pkl', 'rb') as f:
    result_64parcels = pickle.load(f)
    
with open(pkl_path / 'sim_result_20parcels_80split.pkl', 'rb') as f:
    result_20parcels_80split = pickle.load(f)

with open(pkl_path / 'sim_result_36parcels_80split.pkl', 'rb') as f:
    result_36parcels_80split = pickle.load(f)
    
with open(pkl_path / 'sim_result_64parcels_80split.pkl', 'rb') as f:
    result_64parcels_80split = pickle.load(f)
```

# Results for equal sized groups

```python
print_result(result_20parcels, 20)
```

```python
print_result(result_36parcels, 36)
```

```python
print_result(result_64parcels, 64)
```

# Results for unequal sized groups (80% controls)

```python
print_result(result_20parcels_80split, 20)
```

```python
print_result(result_36parcels_80split, 36)
```

```python
print_result(result_64parcels_80split, 64)
```

# Plots

```python
plt.figure(figsize=(10, 6))
plt.plot(result_20parcels['N'], result_20parcels['mean_sensitivity'], marker='o', linestyle='-', color='b', label='20 parcels')
plt.plot(result_36parcels['N'], result_36parcels['mean_sensitivity'], marker='o', linestyle='-', color='r', label='36 parcels')
plt.plot(result_64parcels['N'], result_64parcels['mean_sensitivity'], marker='o', linestyle='-', color='g', label='64 parcels')
#plt.plot(result_20parcels['N'], result_20parcels['mean_specificity'], marker='s', linestyle='-', color='r', label='Mean Specificity')

# Add titles and labels
plt.title('Mean sensitivity vs. N for equal sized groups')
plt.xlabel('N participants')
plt.ylabel('Sensitivity')
plt.ylim(0.05, 1)
plt.grid(True)
plt.legend()

# Show the plot
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.plot(result_20parcels_80split['N'], result_20parcels_80split['mean_sensitivity'], marker='o', linestyle='-', color='b', label='20 parcels')
plt.plot(result_36parcels_80split['N'], result_36parcels_80split['mean_sensitivity'], marker='o', linestyle='-', color='r', label='36 parcels')
plt.plot(result_64parcels_80split['N'], result_64parcels_80split['mean_sensitivity'], marker='o', linestyle='-', color='g', label='64 parcels')
#plt.plot(result_20parcels['N'], result_20parcels['mean_specificity'], marker='s', linestyle='-', color='r', label='Mean Specificity')

# Add titles and labels
plt.title('Mean sensitivity vs. N for unequal sized groups: 80% control')
plt.xlabel('N participants')
plt.ylabel('Sensitivity')
plt.ylim(0.05, 1)
plt.grid(True)
plt.legend()

# Show the plot
plt.show()
```
