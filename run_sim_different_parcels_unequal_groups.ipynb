```python
def calculate_proportion(N_cases, N_controls):
    proportion = N_controls / (N_cases + N_controls)
    return proportion
```

```python
def simulation_loop(N_values, conn_df):
    # Loop through the values of N and run simulation with specified parameters
    result_dict = {"N": [], "mean_sensitivity": [], "mean_specificity": []}
    for N in N_values:
        mean_sens, mean_spef = sim.run_multiple_simulation(conn_df, N=N, pi=0.20, 
                                                       d=0.3, proportion_split=0.8, q=0.1, num_sample=100)
        #print(f"Simulation ran for N={N}.")
        
        # Append results to the dictionary
        result_dict["N"].append(N)
        result_dict["mean_sensitivity"].append(mean_sens)
        result_dict["mean_specificity"].append(mean_spef)
    
    return result_dict
```

```python
# Path to a .csv file with connectomes in upper triangular form
path_conn_20parcels = "/home/neuromod/ad_sz/data/abide/abide1and2_controls_20parcels.csv"
path_conn_36parcels = "/home/neuromod/ad_sz/data/abide/abide1and2_controls_36parcels.csv"
path_conn_64parcels = "/home/neuromod/ad_sz/data/abide/abide1and2_controls_64parcels.csv"
```

```python
# Load control connectomes from ABIDE
conn_20parcels_df = pd.read_csv(path_conn_20parcels)
conn_36parcels_df = pd.read_csv(path_conn_36parcels)
conn_64parcels_df = pd.read_csv(path_conn_64parcels)
```

```python
# Create a range of N values
N_values = range(300, 901, 50)
```

```python
result_20parcels_80split = simulation_loop(N_values, conn_20parcels_df)
result_36parcels_80split = simulation_loop(N_values, conn_36parcels_df)
result_64parcels_80split = simulation_loop(N_values, conn_64parcels_df)
```
